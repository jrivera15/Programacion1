#Elementos básicos de Python

##Reglas para los identificadores

Para crear un identificador, se deben tomar en cuenta las siguientes reglas:

* Únicamente pueden utilizarse letras del alfabeto (a...z, A...Z), dígitos (0...9) y/o el carácter de subrayado (_). A partir de Python 3, puede incluirse la letra ñ, Ñ en el alfabeto y vocales acentuadas.
* Puede iniciarse con letra o el carácter de subrayado (es decir, no puede iniciarse un identificador con un dígito).
* No deben ser palabras reservadas del lenguaje python.
* Python es sensible a mayúsculas y minúsculas.
* Cuando empiezan con el caracter de subrayado, generalmente tienen un propósito especial.


| VÁLIDO  | NO VÁLIDO  |
|---|---|
|suma_total|suma-total |
|  sumaTotal | suma total  |
| cuenta4  | 4cuenta  |
| _42  | 42  |
| SUMA_TOTAL | $UMA_TOTAL|
| nombre | 'nombre  |
| for_var | for|

##Lista de palabras reservadas en Python

```python
import keyword
```

```python
for e in keyword.kwlist:
  print(e)
```

Para saber si una palabra es reservada

```python
print(keyword.iskeyword("for"))      # True
print(keyword.iskeyword("nombre"))   # False
```

Además, Python tiene soft keywords, que son palabras especiales solo en ciertos contextos. Podemos obtener un listado de estas palabras usando el módulo `keyword`.

Una palabra reservada normal nunca puede usarse como identificador. Una soft keyword solo se vuelve especial cuando aparece en una estructura particular del lenguaje. Fuera de ese contexto, puede usarse como nombre, aunque no siempre sea recomendable.

```python
keyword.softkwlist
```

## Expresiones

Una expresión es el tipo más básico de instrucción de programación en el lenguaje. Las expresiones constan de valores (como 2 y 3) y operadores (como +), y siempre pueden evaluarse (es decir, reducirse) a un único valor. Esto significa que puedes utilizar expresiones en cualquier lugar del código de Python donde también podrías usar un valor:


```python
2 + 3
```

Un único valor sin operadores también se considera una expresión, aunque se evalúa como sí mismo:


```python
5
```

##Tipos de datos



En Python no es necesario especificar el tipo de dato al definir una variable. Python es de tipado dinámico

Esto significa que:

* Las variables no tienen un tipo fijo, sino que el tipo depende del valor que le asignas.

* El intérprete detecta automáticamente el tipo según el valor.

Sin embargo, hay varios tipos de datos básicos que se usan todo el tiempo. A continuación una tabla con los tipos de datos que aprenderás a usar.

| Tipo | Descripción |
|------|-------------|
| int | Número entero |
| float | Número decimal (flotante) |
| str | Cadena de texto (string) |
| bool | 	Booleano (verdadero/falso) |
| list | Lista ordenada |
| tuple | Tupla (como lista, pero fija) |
| dict | Diccionario (clave-valor) |
| set | Conjunto (sin duplicados) |
| None | Nulo (como null en otros lenguajes) |

```python
nombre = "hola"
type(nombre)
```

```python
type(2+3)
```

En ocasiones tendrás que utilizar la conversión de tipos (type casting) para asignar el tipo correcto de datos que se requiere.

```python
#Algunos ejemplos de conversión de tipo
x1 = int("10")     # de string a int
x2 = int(3.7)      # de float a int (trunca, no redondea)
x3 = float("3.14") # de string a float
x4 = float(5)      # de int a float (5.0)
x5 = str(42)       # de int a string
x6 = str(3.14)     # de float a string
```

Si intentas convertir algo que no tiene sentido, da error:

```python
int("hola")  # ValueError
```

#Operadores

## Operadores aritméticos



| Operador | Significado      | Ejemplo  |
| -------- | ---------------- | -------- |
| `+`      | suma             | `a + b`  |
| `-`      | resta            | `a - b`  |
| `*`      | multiplicación   | `a * b`  |
| `/`      | división real    | `a / b`  |
| `//`     | división entera  | `a // b` |
| `%`      | módulo o residuo | `a % b`  |
| `**`     | potencia         | `a ** b` |


```python
a = 10
b = 3

print(a + b)   # 13
print(a / b)   # 3.333...
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
```

##Operadores relacionales

| Operador | Significado       | Ejemplo  |
| -------- | ----------------- | -------- |
| `==`     | igual que         | `a == b` |
| `!=`     | diferente de      | `a != b` |
| `<`      | menor que         | `a < b`  |
| `>`      | mayor que         | `a > b`  |
| `<=`     | menor o igual que | `a <= b` |
| `>=`     | mayor o igual que | `a >= b` |


```python
calificacion = 85

if calificacion>=0 and calificacion <= 100:
    print("Calificación válida")
```

##Operadores lógicos

Ejemplo
Separar un número en millares, centenas, decenas y unidades

```python
numero = int(input("Escribr un número entero: "))
millares = numero // 1000
centenas = (numero // 100) % 10
decenas = (numero // 10) % 10
unidades = numero % 10
print(f"Millares: {millares}")
print(f"Centenas: {centenas}")
print(f"Decenas: {decenas}")
print(f"Unidades: {unidades}")
```

```python
s = "hello"
```

```python
type(s)
```

```python
print( f"La cadena {s} tiene { len(s) } caracteres" )
```

```python
numcaracteres = len(s)
print( f"La cadena {s} tiene { numcaracteres } caracteres" )
```

```python
print(f"Primer caracter de {s}={s[0]}")     # h
print(f"Segundo caracter de {s}={s[1]}")    # e

print(f"Último caracter de {s}={s[len(s)-1]}")     # o
print(f"Último caracter de {s}={s[numcaracteres-1]}")     # o
print(f"Último caracter de {s}={s[-1]}")    # o
```

```python
# Slicing de cadenas
s = "hello world"
print(s[0:5]) # hello
print(s[6:]) # world
```

```python
# Métodos útiles
s = "Python is fun!"
print(s.lower())                # python is fun!
print(s.upper())                # PYTHON IS FUN!
print(s.find("fun"))            # 10
print(s.replace("fun", "cool")) # Python is cool!
```

```python
# Métodos split() y join()
text = "one,two,three"
parts = text.split(",")         # ['one', 'two', 'three']
joined = "-".join(parts)        # one-two-three
```

```python
print(text)
print(parts)
print(joined)
```

```python
cstr = "Me encanta estudiar en la UNISON"
print ("La cadena alineada al centro  : ")
print (cstr.center(40), "\n")

# Printing the center aligned string with fillchr
print("Centrada con cartacter de relleno: ")
print(cstr.center(40, '#'))
print()
# Printing the left aligned string with "-" padding
print("Alineada a la izquierda con caracter de relleno: ")
print(cstr.ljust(40, '-'))
print()
# Printing the right aligned string with "-" padding
print("Alineada a la derecha con caracter de relleno: ")
print(cstr.rjust(40, '-'))
```

La estructura general es

`objeto.metodo()`

o, si necesita datos adicionales:

`objeto.metodo(argumentos)`


**Un método es una acción que un objeto sabe hacer.**
