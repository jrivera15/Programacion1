#Python 🐍

**Python** es un lenguaje de programación con las siguientes características:
<br><br>

*   **De alto nivel**: el código es fácil de leer y escribir por humanos, ya que utiliza palabras en inglés (como if, for, print) en lugar de usar ceros y unos (como el lenguaje máquina) o códigos complejos que solo entiende el procesador de la computadora (como el lenguaje ensamblador).

* **Interpretado:** el programa se ejecuta línea por línea directamente. No necesitas esperar a transformar todo tu código en un archivo ejecutable antes de probarlo; una herramienta interna llamada "intérprete" lee cada instrucción y la traduce a la computadora al instante.

* **De propósito general:** sirve para crear casi cualquier cosa. No está limitado a una sola tarea; con él puedes programar desde algo muy básico como automatizar una tarea repetitiva, hasta crear una página web o un videojuego, incluso un sistema de inteligencia artificial o un robot.

<br>

Creado por **Guido van Rossum** en **1991**. Su filosofía prioriza la legibilidad del código y una sintaxis sencilla similar al lenguaje humano.

##[Funciones de orden superior]

Objetivos
- Comprender qué son las funciones de orden superior (HOF)
- Usar `map()`, `filter()`, `reduce()` y `lambda`
- Pasar funciones como argumentos y devolverlos
- Aplicar la programación funcional a problemas reales

Definiciones

Funciones de orden superior: funciones que toman otras funciones como argumentos o las devuelven como resultados.

Funciones lambda: son funciones pequeñas y anónimas definidas mediante la palabra clave `lambda`. Se suelen usar para operaciones cortas y sencillas que pueden expresarse en una sola línea.

Operación `map`: la operación `map` aplica una función unaria a cada elemento de una secuencia y devuelve una secuencia nueva que contiene los resultados en el mismo orden.

Operación `filter`: la operación `filter` le aplica un predicado unario a cada elemento de una secuencia de entrada y produce una secuencia de salida que contiene solo aquellos elementos que cumplen con el predicado.

Operación `reduce`: la operación `reduce` combina los elementos de una secuencia utilizando una función binaria.
